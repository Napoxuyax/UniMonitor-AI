Демонстрационные данные синтетические, не официальная статистика. Для реального эксперимента замените путь и источник. Запуск из корня проекта.

```python
from pathlib import Path
from export_models import export_models
import pandas as pd
import tempfile
output = Path(tempfile.mkdtemp()) / "models"
meta = export_models(Path("data/demo.csv"), output, "Синтетический набор make_demo.py", "demo")
pd.DataFrame(meta["models"])
```

```python
import joblib
import numpy as np
from core import FEATURES
test = pd.read_csv(output / "test.csv")
pred = pd.read_csv(output / "predictions.csv")
for i in (1, 2):
    model = joblib.load(output / f"model{i}.joblib")
    np.testing.assert_allclose(np.clip(model.predict(test[FEATURES]), 0, 100), pred[f"model{i}"])
print("Прогнозы совпадают")
```
